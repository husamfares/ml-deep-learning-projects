```python
all_cols=dataset.columns
i = len(all_cols)-1
Investment_Number=[]
while(i>=0) :
  x = all_cols[i].split(" ")
  size=len(x)

  j=0
  while(size > j):
    if(x[j] == "Asia"):
      Investment_Number.append(all_cols[i])
    j+=1

  i-=1



colors = {"Inward": "tab:blue", "Outward": "tab:orange"}
fig, ax = plt.subplots(figsize=(10,5))

for name in Investment_Number:
       direction = "Inward" if "Inward" in name else "Outward"
       sns.lineplot(x="Year", y=name, data=dataset,
                 label=name.replace("Asia ", ""),
                 color=colors[direction], marker="o")

ax.set_title("Economy Chart")
ax.set_xlabel("Year")
ax.set_ylabel("Value of M&A Deals")
ax.legend(title="Series")
fig.tight_layout()
plt.show()


```

```python
# --- Build strong_cols FIRST (from your raw dataset) ---
# Regex groups that match the ONS column names
pat_in_val  = r"^M&A : Inward : Value of acquisitions"
pat_out_val = r"^M&A : Outward : Value of acquisitions"
pat_dom_val = r"^M&A : Domestic : .*(Value|Total payments)"
pat_in_num  = r"^M&A : Inward : Number of acquisitions"
pat_out_num = r"^M&A : Outward : Number of acquisitions"

# Collect columns
in_val_cols   = dataset.filter(regex=pat_in_val).columns
out_val_cols  = dataset.filter(regex=pat_out_val).columns
dom_val_cols  = dataset.filter(regex=pat_dom_val).columns
in_num_cols   = dataset.filter(regex=pat_in_num).columns
out_num_cols  = dataset.filter(regex=pat_out_num).columns

# Optional regional VALUE totals
asia_val_cols   = dataset.filter(regex=r"^Asia .*Acquisitions Value").columns
africa_val_cols = dataset.filter(regex=r"^Africa .*Acquisitions Value").columns

# Master list used below
strong_cols = (list(in_val_cols) + list(out_val_cols) + list(dom_val_cols) +
               list(in_num_cols) + list(out_num_cols) +
               list(asia_val_cols) + list(africa_val_cols))

print("Strong cols found:", len(strong_cols))
```

```python
nan_per_col = dataset[strong_cols].isna().sum().sort_values()  # fewest NaNs first
core_cols = nan_per_col.index[:31].tolist()

vary = dataset[core_cols].nunique(dropna=True)
core_cols = [c for c in core_cols if vary[c] > 1]
print(len(core_cols))
print("Core columns:", core_cols)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X , y , test_size=0.2 ,shuffle=False, random_state=0)
```

```python
X_train
```

```python
from sklearn.impute import KNNImputer
impute = KNNImputer(n_neighbors=3, weights="distance")
X_train[:, :-1] = impute.fit_transform(X_train[:, :-1])
X_test[:, :-1] = impute.transform(X_test[:, :-1])
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train[:, :-1]= sc.fit_transform(X_train[:, :-1])
X_test[:, :-1] = sc.transform(X_test[:, :-1])
```

```python
from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier(criterion = 'entropy',max_depth=3 , min_samples_leaf=3, random_state = 0)
classifier.fit(X_train, y_train)
```

```python
y_pred = classifier.predict(X_test)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
print(cm)
ac = accuracy_score(y_test, y_pred)
print(ac)
```
